```python
import numpy as np
import pandas as pd
```

```python
def load_data(filename):
    data=pd.read_csv(filename)
    return data
df=load_data('students.csv')
print(df)
```

```python
def clean_data(df):
    df = df.copy()
    df["Name"] = df["Name"].str.strip()
    df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
    df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
    df = df.dropna(subset=["Name", "Age", "Score"])
    df = df[
        df["Age"].between(5, 100)
        & df["Score"].between(0, 100)
        & (df["Name"] != "")
    ]

    df = df.drop_duplicates()

    df["Age"] = df["Age"].astype(int)

    return df.reset_index(drop=True)


cleaned_df = clean_data(df)

print("Cleaned data:")
print(cleaned_df)
```

```python
def process_data(df):
    if df.empty:
        return None

    scores = df["Score"].to_numpy(dtype=float)
    ages = df["Age"].to_numpy(dtype=int)

    return {
        "student_count": len(df),
        "average_score": np.mean(scores),
        "highest_score": np.max(scores),
        "lowest_score": np.min(scores),
        "average_age": np.mean(ages)
    }


results = process_data(cleaned_df)
print(results)
```

```python
def display_summary(results):
    if results is None:
        print("No valid records to process.")
        return

    print("\n===== STUDENT PERFORMANCE REPORT =====")
    print(f"Number of students: {results['student_count']}")
    print(f"Average score: {results['average_score']:.2f}")
    print(f"Highest score: {results['highest_score']:.2f}")
    print(f"Lowest score: {results['lowest_score']:.2f}")
    print(f"Average age: {results['average_age']:.2f}")
    print("=======================================")


display_summary(results)
```
